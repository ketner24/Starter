```python
import numpy as np
import cvxpy as cvx
import scipy.sparse as sparse
import time
```

**Continuation of In-Class Example**: A barrage of 100 missiles is incoming. You have 100 interceptors and need to assign them to the incoming missiles. The matrix $C \in  \mathbb{R}^{100 \times 100}$ is such that $C_{ij}$ gives the initial distance between missile $i$ and interceptor $j$. Find an assignment of interceptors to missiles that minimizes the cumulative sum of distances each interceptor must travel.

```python
#generate synthetic problem data
np.random.seed(0)
n = 100
C = np.random.uniform(low=0.1, high=100, size=(100,100))
```

```python
#first attempt. Scalar form
#I'm declaring X as a 2D variable
#with x_{ij} denoting missile i and intercepter j
X = cvx.Variable((n,n), nonneg=True)
objective = cvx.Minimize(cvx.sum(C[i,j]*X[i,j] for i in range(n) for j in range(n)))
#first we'll do row column sums
constraints = [cvx.sum(X[i,j] for i in range(n)) == 1. for j in range(n)]
#now we'll do row sums
constraints += [cvx.sum(X[i,j] for j in range(n)) == 1. for i in range(n)]
#finally we'll add the upper bounds to the variables
constraints += [X[i,j] <= 1. for i in range(n) for j in range(n)]
prob = cvx.Problem(objective, constraints)
start = time.time()
prob.solve(solver='HIGHS', verbose=True)
print(f"Took {time.time() - start} seconds to compile and solve")
```

```python
#second attempt. Something between matrix and scalar form
X = cvx.Variable((n,n), nonneg=True)
objective = cvx.Minimize(cvx.sum(cvx.multiply(C, X)))
#first we'll do row column sums
constraints = [cvx.sum(X[:,j]) == 1. for j in range(n)]
#now we'll do row sums
constraints += [cvx.sum(X[i,:]) == 1. for i in range(n)]
#finally we'll add the upper bounds to the variables
constraints += [X <= 1.]
prob = cvx.Problem(objective, constraints)
start = time.time()
prob.solve(solver='HIGHS', verbose=True)
print(f"Took {time.time() - start} seconds to compile and solve")
```

```python
#third attempt, Sparse Matrix/Vector form
c = C.ravel()
x = cvx.Variable(n*n, nonneg=True)
objective = cvx.Minimize(c.T @ x)
#sum over interceptors matrix
sum_int_mat = sparse.coo_array((n, n*n))
for i in range(n):
    sum_int_mat[i,n*i:n*(i+1)] = 1.
#sum over missiles matrix
sum_miss_mat = sparse.coo_array((n, n*n))
for i in range(n):
    sum_miss_mat[i,i::n] = 1.
D = sparse.bmat([[sum_int_mat],[sum_miss_mat]])
constraints = [D @ x == 1., x <= 1.]
prob = cvx.Problem(objective, constraints)
start = time.time()
prob.solve(solver='HIGHS', verbose=True)
print(f"Took {time.time() - start} seconds to compile and solve")
```

```python
#another attempt. This time use bounds instead of x <= 1. constraint
c = C.ravel()
x = cvx.Variable(n*n, bounds=(0., 1.))
objective = cvx.Minimize(c.T @ x)
#sum over interceptors matrix
sum_int_mat = sparse.coo_array((n, n*n))
for i in range(n):
    sum_int_mat[i,n*i:n*(i+1)] = 1.
#sum over missiles matrix
sum_miss_mat = sparse.coo_array((n, n*n))
for i in range(n):
    sum_miss_mat[i,i::n] = 1.
D = sparse.bmat([[sum_int_mat],[sum_miss_mat]])
constraints = [D @ x == 1.]
prob = cvx.Problem(objective, constraints)
start = time.time()
prob.solve(solver='HIGHS', verbose=True)
print(f"Took {time.time() - start} seconds to compile and solve")
```

```python
#here's a pyomo implementation, for comparison
import pyomo.environ as pyo
model = pyo.ConcreteModel()
missiles = range(n)
interceptors = range(n)
model.X = pyo.Var(missiles, interceptors, domain=pyo.NonNegativeReals) #could do domain=pyo.realds, bounds=(0.,1.) and omit the bound_X constraint
model.objective = pyo.Objective(expr=sum(C[i,j]*model.X[i,j] for i in missiles for j in interceptors),
                                sense=pyo.minimize)
model.missile_sum = pyo.Constraint(missiles,
                                   rule=lambda mod, i: sum(mod.X[i,j] for j in interceptors)==1.)
model.interceptor_sum = pyo.Constraint(interceptors,
                                   rule=lambda mod, j: sum(mod.X[i,j] for i in missiles)==1.)
model.bound_X = pyo.Constraint(missiles, interceptors,
                               rule=lambda mod, i, j: mod.X[i,j] <= 1.)
opt = pyo.SolverFactory('appsi_highs')
start = time.time()
opt.solve(model, tee=True) #tee=True is (supposed to be) the verbose option for pyomo. not working here
print(f"Took {time.time() - start} seconds to compile and solve")
print("Optimal objective", pyo.value(model.objective))
```
